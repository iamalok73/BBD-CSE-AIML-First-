# Tokenization
- Tokenization is the process of breaking down a piece of text into smaller units called tokens. These tokens can be words, sentences, or even subwords, depending on the rules applied. Tokenization is a fundamental step in Natural Language Processing (NLP) as it helps in preparing the text for further analysis or processing.

# Handling Contraction

- In Natural Language Processing (NLP), contractions are shortened forms of words or phrases, such as "can't" for "cannot" or "it's" for "it is." Expanding these contractions is a crucial preprocessing step to improve text understanding, maintain context, and enhance the performance of downstream tasks like sentiment analysis, machine translation, and named entity recognition.

```python
pip install contractions
```

```python
import contractions
```

```python
detail = "I'm living with my Dad house. Today I'll not going to school."
new = contractions.fix(detail)

print(detail)
print(new)

print(nltk.word_tokenize(new))
```

# Stop word Removing 

```python
import nltk
nltk.download('stopwords')
nltk.download('punkt')
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
detail = "I'm living with my Dad house. Today I'll not going to school."
new = contractions.fix(detail)

x = (word_tokenize(new))
print(x)

stop_words = set(stopwords.words('english'))

# remove stopwords
filtered = [i for i in x if i.lower() not in stop_words]

print(filtered)
```
