```python

import spacy 
nlp = spacy.load("en_core_web_sm")
```

```python
content = "Elon musk is owner of spaceX."
```

```python
doc = nlp("Elon musk is owner of spaceX.")
for ent in doc.ents:
    print(ent.text, ent.start_char, ent.end_char, ent.label_)
```

```python
doc = nlp("Elon musk is owner of SpacX.")
for ent in doc.ents:
    print(ent.text, ent.label_)
```
