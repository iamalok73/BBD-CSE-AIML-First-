```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\AlokVerma\Desktop\ML DATA\Telco_Cusomer_Churn.csv")
```

```python
df.drop(columns="customerID",inplace=True)
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'],errors="coerce")
```

```python
df.dropna(inplace=True)
```

```python
df = pd.get_dummies(df,columns=['gender',  'Partner', 'Dependents', 
       'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity',
       'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV',
       'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod','Churn'],drop_first=True)
```

```python
X = df.iloc[:,0:30]
```

```python
y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
X_train_sc = sc.fit_transform(X_train)
X_test_sc = sc.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train_sc,y_train)
```

```python
y_pred = model.predict(X_test_sc)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
X_train
```

```python
import keras
```

```python
from keras.layers import Dense
from keras.models import Sequential
```

```python
model  = Sequential()
```

```python
model.add(Dense(units = 10,activation="relu",kernel_initializer="uniform"))
model.add(Dense(units = 10,activation="relu",kernel_initializer="uniform"))
model.add(Dense(units = 10,activation="relu",kernel_initializer="uniform"))
```

```python
model.add(Dense(units = 1,activation="sigmoid",kernel_initializer="uniform"))
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss=keras.losses.BinaryCrossentropy(),
    metrics=[
        keras.metrics.BinaryAccuracy(),
        keras.metrics.FalseNegatives(),
    ],
)
```

```python
abc = model.fit(X_train,y_train,batch_size=10,epochs=50)
```

```python
abc.params
```

```python
accuracy = model.evaluate(X_test,y_test)
```
