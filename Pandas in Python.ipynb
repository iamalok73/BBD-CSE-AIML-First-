```python
data_1 = {
    "id":[1,2,3,4,9,11],
    "Name":["Amit","Sumit","Raju","vinit","ankit","Rahul"] 
}
```

```python
data_2 = {
    "id":[1,2,3,5,6,7,18,19],
    "Marks":[34,67,33,57,89,65,88,56]
}
```

```python
df_1 = pd.DataFrame(data_1)
```

```python
df_2 = pd.DataFrame(data_2)
```

```python
df_1
```

```python
df_2
```

```python
# merge
```

```python
df_3 = pd.merge(df_1,df_2,how="inner",on="id")

df_3
```

```python
df_4 = pd.merge(df_1,df_2,how="left",on="id")

df_4
```

```python
df_5 = pd.merge(df_1,df_2,how="right",on="id")

df_5
```

```python
df_6 = pd.merge(df_1,df_2,how="outer",on="id")

df_6
```
