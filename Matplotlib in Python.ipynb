```python
a = np.linspace(0,10,100)
```

```python
print(a)
```

```python
plt.figure(figsize=(6,4))
plt.plot(a,np.sin(a),"r^")
plt.plot(a,np.cos(a))
plt.xlabel("this is X side")
plt.ylabel("this is y label.")
plt.title("title of this graph")
plt.legend(a)
plt.show()
```
